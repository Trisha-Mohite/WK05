### Some More Questions:

- What's the tallest work in the collection ?
- What's the heaviest ?
- Widest ?

These can be answered by iterating through the list of artworks and keeping the artwork with the largest value seen in a separate variable.

```python
# NOTE: for each of these, assign the first artwork as the tallest/heaviest/widest
#       and then iterate over all works and find taller, heavier, wider ones

# TODO: Tallest ?
tallest = artworks[0]

for artwork in artworks:
    if artwork["Height (cm)"] > tallest["Height (cm)"]:
        tallest = artwork

print(tallest)


# TODO: Heaviest ?

heaviest = artworks[0]

for artwork in artworks:
    if artwork["Weight (kg)"] > heaviest["Weight (kg)"]:
        tallest = artwork

print(heaviest)

# TODO: Widest ?

widest = artworks[0]

for artwork in artworks:
    if artwork["Width (cm)"] > widest["Width (cm)"]:
        widest = artwork

print(widest)
```
